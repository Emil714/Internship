```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.cluster import KMeans
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import silhouette_score
```

```python
df = pd.read_csv("train (1).csv")

df1 = df.copy()

df1.head(10)
```

```python
gender_map = {
    'Male': 1,
    'Female': 0
}

df1['Gender'] = df1['Gender'].map(gender_map)

df1
```

```python
df1.isna().any().any()
```

```python
features = ['CustomerID', 'Gender','Age', 'Annual Income (k$)', 'Spending Score (1-100)']
X = df1[features].copy()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df1)

X_scaled_df = pd.DataFrame(X_scaled, columns=features)
X_scaled_df.describe()
```

```python
wcss = []
score = []
Ks = range(2, 11)

for k in Ks:
    kmeans = KMeans(n_clusters=k, init = "k-means++", n_init=10, random_state=42)
    preds = kmeans.fit_predict(X_scaled)
    wcss.append(kmeans.inertia_)
    score.append(silhouette_score(X_scaled, preds))
    
for k, w, s in zip(Ks, wcss, score):
    print(f"K={k}: WCSS={w:.2f}, Silhouette={s:.3f}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(list(Ks), wcss, marker='o', color='steelblue')
axes[0].set_xlabel('Number of clusters (K)')
axes[0].set_ylabel('WCSS (Inertia)')
axes[0].set_title('Elbow Method')

axes[1].plot(list(Ks), score, marker='o', color='indianred')
axes[1].set_xlabel('Number of clusters (K)')
axes[1].set_ylabel('Silhouette Score')
axes[1].set_title('Silhouette Score vs K')

plt.tight_layout()
plt.show()
```

```python
kmeans = KMeans(n_clusters=5, init="k-means++", n_init=10, random_state=42)
cl_preds = kmeans.fit_predict(X_scaled)

cl_preds
```
