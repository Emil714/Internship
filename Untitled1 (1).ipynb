```python
df_train['Title'] = df_train['Name'].str.extract(r',\s*(\w+\.)\s*')

title_mapping = {
    'Miss.': 0,
    'Mrs.': 1,
    'Master.': 4,
    'Mr.': 5
}

df_train['Title_encoded'] = df_train['Title'].map(title_mapping)

df_train = df_train.drop(columns=['Name', 'Title'])

df_train['Title_encoded'] = df_train['Title_encoded'].fillna(df_train['Title_encoded'].median())

df_train = pd.get_dummies(df_train, columns=['Title_encoded'], prefix='Title')
```

```python
df_train
```

```python
df_train = pd.get_dummies(df_train, columns=['Embarked'], prefix='Embarked')
df_train
```

```python
sex_mapping = {
    'male': 1,
    'female': 0
}

df_train['Sex'] = df_train['Sex'].map(sex_mapping)

df_train
```

```python
df_train = df_train.drop('Ticket', axis=1)
df_train
```

```python
X = df_train.drop('Survived', axis=1)
y = df_train['Survived']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.2)
```

```python
import xgboost as xgb
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import cross_validate

knn = KNeighborsRegressor(n_neighbors=15)

pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('knn', knn)
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

results = cross_validate(
    pipe, X_train, y_train,
    cv=cv,
    scoring=['accuracy', 'f1', 'roc_auc'],  
    return_train_score=True,
    n_jobs=-1
)

pipe.fit(X_train, y_train)
```

```python
### Замерим accuracy

print(f"Качество на трейне: {np.mean(pipe.predict(X_train) == y_train).round(3)}")

print(f"Качество на тесте: {np.mean(pipe.predict(X_test) == y_test).round(3)}")
```

```python
from sklearn.svm import LinearSVC


pipe_svm = Pipeline([("scaler_svm", StandardScaler()),
                     ("SVM", LinearSVC())])


pipe_svm.fit(X_train, y_train)

pipe_svm.score(X_test, y_test)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score


print(f'Accuracy равно: {accuracy_score(y_test, pipe_svm.predict(X_test)):.3f}' )
```

```python
boost = xgb.XGBClassifier(
    n_estimators=200,        # количество деревьев (boosting rounds)
    max_depth=6,              # максимальная глубина каждого дерева
    learning_rate=0.1,        # скорость обучения (шаг градиентного спуска)
    subsample=0.8,            # доля объектов для каждого дерева
    colsample_bytree=0.8,     # доля признаков для каждого дерева
    random_state=42,
    eval_metric='logloss'     # метрика для отслеживания качеств
)

new_pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('xgb', boost)
])

new_pipe.fit(X_train, y_train)
```

```python
print(f'Accuracy равно: {accuracy_score(y_test, new_pipe.predict(X_test)):.3f}' )
```
